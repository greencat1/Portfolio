### Imports

```python
import json
import numpy as np
import faiss
import requests
from sentence_transformers import CrossEncoder

from sentence_transformers import SentenceTransformer
```

### Load FAISS + metadata

```python
INDEX_PATH = "../data/embeddings/faiss.index"
META_PATH = "../data/embeddings/metadata.json"

index = faiss.read_index(INDEX_PATH)

with open(META_PATH, "r", encoding="utf-8") as f:
    chunks = json.load(f)

print(f"Loaded {len(chunks)} chunks")
```

### Load embedding model

```python
model = SentenceTransformer("BAAI/bge-small-en-v1.5")
```

### Retrieval

```python
def search(query, k=20):
    q_emb = model.encode([query]).astype("float32")
    distances, indices = index.search(q_emb, k)
    return [chunks[i] for i in indices[0]]
```

### Deduplication

```python
def unique_results(results, max_results=5):
    seen = set()
    unique = []

    for r in results:
        text = r["text"]

        if text[:200] not in seen:
            unique.append(r)
            seen.add(text[:200])

        if len(unique) >= max_results:
            break

    return unique
```

### Build context

```python
def build_context(results):
    return "\n\n".join([r["text"] for r in results])
```

### Prompt (VERY IMPORTANT for local models)

```python
def build_prompt(query, context):
    return f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>
You are an ML project expert answering questions about a technical portfolio.

CRITICAL RULES:
1. Answer ONLY using information from the context below
2. If the context contains a specific answer — give it directly and concisely
3. If multiple options are mentioned but one is identified as "best" or "final" — state that one
4. Use exact numbers, metrics, and model names from the context
5. Keep answers to 1-3 sentences maximum
6. If the answer is NOT in the context, say exactly: "I don't know"
7. Never say "the documentation does not specify" — either give the answer or say "I don't know"

<|eot_id|><|start_header_id|>user<|end_header_id|>
CONTEXT:
---
{context}
---

QUESTION: {query}

ANSWER (concise, factual, 1-3 sentences):
<|eot_id|><|start_header_id|>assistant<|end_header_id|>"""
```

### Call local LLM (Ollama)

```python
def ask_llm(prompt):
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "phi3:mini",
            "prompt": prompt,
            "stream": False,
            "options": {
                "num_predict": 512,
                "temperature": 0.1
            }
        }
    )
    return response.json()["response"]
```

### Full RAG pipeline

```python
def rag(query):
    # 1. Retrieve
    results = search(query, k=20)
    
    # 2. Deduplicate
    results = unique_results(results, max_results=5)
    
    # 3. Build context
    context = build_context(results)
    
    # 4. Prompt
    prompt = build_prompt(query, context)
    
    # 5. LLM
    answer = ask_llm(prompt)
    
    return answer, results
```

## Summary 

This notebook implements a **complete RAG (Retrieval-Augmented Generation) pipeline** for querying an ML portfolio.

---

###  Main Components

| Component | What it does |
|-----------|---------------|
| **FAISS + metadata** | Loads pre-computed embeddings and 8,361 text chunks |
| **BGE embedding model** | Encodes user queries into vectors (`BAAI/bge-small-en-v1.5`) |
| **Semantic search** | Finds top-20 relevant chunks using cosine similarity |
| **Deduplication** | Removes duplicate chunks (keeps top-5 unique) |
| **LLM (phi3:mini)** | Generates concise answers via local Ollama |
| **Strict prompt** | Forces factual answers from context only |

---

###  Pipeline Flow
Query → Embed → FAISS Search (k=20) → Deduplicate (top-5) → Build Context → LLM (phi3:mini) → Answer


---

###  Key Functions

| Function | Purpose |
|----------|---------|
| `search(query, k=20)` | Retrieves chunks from FAISS index |
| `unique_results(results, max_results=5)` | Removes near-duplicate chunks |
| `build_context(results)` | Joins chunks into a single context string |
| `build_prompt(query, context)` | Creates a strict, fact-based prompt |
| `ask_llm(prompt)` | Calls local Ollama with phi3:mini |
| `rag(query)` | Orchestrates the full pipeline |

---

###  Prompt Rules (Critical)

1. Answer ONLY using context
2. Give direct, concise answers
3. If "best" or "final" is mentioned — state that one
4. Use exact numbers and model names
5. Keep answers to 1-3 sentences
6. If not in context → "I don't know"
7. Never say "the documentation does not specify"

---

###  Summary

This notebook turns the portfolio into a **question-answering system**. User asks a question → system finds relevant chunks → LLM answers based ONLY on those chunks. Results are factual, concise, and grounded in the actual portfolio content.
